```python
import pandas as pd
```

```python
excel_file = pd.ExcelFile('chp_data.xlsx')
```

```python
print("Your tabs are:")
print(excel_file.sheet_names)
```

```python
first_sheet_name = excel_file.sheet_names[0]
combined_table = pd.read_excel(excel_file, sheet_name=first_sheet_name)
print(f"Started with {first_sheet_name}")
```

```python
for i in range(1, len(excel_file.sheet_names)):
    sheet_name = excel_file.sheet_names[i]
    next_table = pd.read_excel(excel_file, sheet_name=sheet_name)
    combined_table = combined_table.merge(
        next_table,
        on = 'Pursuit Number',
        how = 'outer',
        suffixes = ('', f'_{sheet_name}')
    )
```

```python
print(f"Added {sheet_name}")
```

```python
combined_table.to_excel('combined_output.xlsx', index = False)
```
