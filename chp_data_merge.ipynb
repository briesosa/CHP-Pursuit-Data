```python
# Set the option to display all columns
pd.set_option('display.max_columns', None)
```

```python
result = main_file.copy()
result = result.merge(suspect_year, on=shared_column, how="left", suffixes=('', '_suspect'))
result = result.merge(total_injury, on=shared_column, how="left", suffixes=('', '_injury'))
result = result.merge(initiating_year, on=shared_column, how="left", suffixes=('', '_initiating'))
result = result.merge(terminating_info, on=shared_column, how="left", suffixes=('', '_terminating'))
#result = result.merge(other_violation, on=shared_column, how="left", suffixes=('', '_other'))
```
